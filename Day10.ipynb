1. Class Student, attributes name & roll, two objects

```python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
    def display(self):
        print("Name:", self.name)
        print("Roll:", self.roll)
s1 = Student("Shreya", 101)
s2 = Student("Satyam", 102)
s1.display()
s2.display()
```

2. Constructor se employee name & salary initialize

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def show(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
e = Employee("Shreya", 50000)
e.show()
```

3. Class attribute & instance attribute

```python
class Company:
    company_name = "TCS"   
    def __init__(self, emp_name):
        self.emp_name = emp_name   
c1 = Company("Shreya")
c2 = Company("Satyam")
print(c1.company_name)
print(c1.emp_name)
print(c2.company_name)
print(c2.emp_name)
```

4. Parent class Person → Child class Student

```python
class Person:
    def display(self):
        print("I am a Person")
class Student(Person):
    pass
s = Student()
s.display()
```

5. Method overriding

```python
class Parent:
    def show(self):
        print("Parent class")
class Child(Parent):
    def show(self):
        print("Child class")
c = Child()
c.show()
```

6. Private variable access using public method

```python
class Bank:
    def __init__(self):
        self.__balance = 100000 
    def show_balance(self):
        print("Balance:", self.__balance)
b = Bank()
b.show_balance()
```

7. Protected member access in child class

```python
class Parent:
    def __init__(self):
        self._value = 500
class Child(Parent):
    def show(self):
        print("Value:", self._value)
c = Child()
c.show()
```

8. Runtime polymorphism using loop

```python
class Dog:
    def speak(self):
        print("Dog barks")
class Cat:
    def speak(self):
        print("Cat meo")
animals = [Dog(), Cat()]
for a in animals:
    a.speak()
```

9. IS-A vs HAS-A relationship

```python
class Animal:
    def eat(self):
        print("Animal eats")
class Dog(Animal):
    pass
class Engine:
    def start(self):
        print("Engine started")
class Car:
    def __init__(self):
        self.engine = Engine()
d = Dog()
d.eat()
c = Car()
c.engine.start()
```

10. super() use karke parent constructor call

```python
class Parent:
    def __init__(self, name):
        self.name = name
class Child(Parent):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)
c = Child("Shreya", 20)
c.show()
```

11. Diamond Problem (MRO demonstration)

```python
class A:
    def show(self):
        print("Class A")
class B(A):
    def show(self):
        print("Class B")
class C(A):
    def show(self):
        print("Class C")
class D(B, C):
    pass
d = D()
d.show()
print(D.mro())
```

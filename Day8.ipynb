1. Create a file and write 5 lines using "w" mode

```python
f = open("data.txt", "w")
f.write("Line 1\n")
f.write("Line 2\n")
f.write("Line 3\n")
f.write("Line 4\n")
f.write("Line 5\n")
f.close()
```

2. Read the file using "r" mode

```python
f = open("data.txt", "r")
content = f.read()
print(content)
f.close()
```

3. Append a new line using "a" mode

```python
f = open("data.txt", "a")
f.write("This is appended line\n")
f.close()
```

4. Use "r+" mode to read and write in same file

```python
f = open("data.txt", "r+")
print(f.read())
f.write("\nWritten using r+ mode")
f.close()
```

5. Print cursor position using tell()

```python
f = open("data.txt", "r")
print("Cursor position:", f.tell())
f.read(10)
print("Cursor after reading 10 chars:", f.tell())
f.close()
```

6. Move cursor using seek() and re-read content

```python
f = open("data.txt", "r")
print(f.read(15))
f.seek(0)   
print("After seek:")
print(f.read())
f.close()
```

7. Store student data in a JSON file

```python
import json
student = {
    "id": 1,
    "name": "Shreya",
    "age": 20,
    "course": "BCA"
}
with open("student.json", "w") as f:
    json.dump(student, f)
```

8. Read JSON file and print student details

```python
import json
with open("student.json", "r") as f:
    data = json.load(f)
print("ID:", data["id"])
print("Name:", data["name"])
print("Age:", data["age"])
print("Course:", data["course"])
```
